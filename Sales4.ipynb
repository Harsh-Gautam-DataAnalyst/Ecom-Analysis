```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
data=pd.read_csv('C:/Users/hp/Downloads/sales3.csv')
```

```python
data.head()
```

```python
df=data.copy()
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df=df.drop(columns={'Social_Media_Influence','Engagement_with_Ads'})
```

```python
df.info()
```

```python
df['Time_of_Purchase']=pd.to_datetime(df['Time_of_Purchase'])
df['Month']=df['Time_of_Purchase'].dt.month
```

```python
df['Purchase_Amount'] = df['Purchase_Amount'].replace(r'[\$\s]', '', regex=True)
df['Purchase_Amount'] = df['Purchase_Amount'].astype(float)
```

```python
df.head()
```

```python
df.describe()
```

```python
df.to_csv('cleaned_sales4.csv')
```

# EDA

```python
df.columns
```

```python
#sales by Gender
x=df.groupby(['Gender'])['Purchase_Amount'].sum().sort_values(ascending=False).reset_index()
print(x)
sns.barplot(data=x,x='Gender',y='Purchase_Amount')
sns.set(rc={'figure.figsize':(20,5)})
plt.show()
```

```python
#sales by income level
q=df.groupby(['Income_Level'])['Purchase_Amount'].sum().sort_values(ascending=False).reset_index()
sns.barplot(x='Income_Level',y='Purchase_Amount',data=q)
print(q)
sns.set(rc={'figure.figsize':(3,2)})
plt.show()
```

```python
#sales by Marital status
ms=df.groupby(['Marital_Status'])['Purchase_Amount'].sum().sort_values(ascending=False).reset_index()
print(ms)

plt.pie(ms['Purchase_Amount'],labels=ms['Marital_Status'],startangle=140,autopct='%1.1f%%')
plt.show()
```

```python
#sales by Education Level
el=df.groupby(['Education_Level'])['Purchase_Amount'].sum().sort_values(ascending=False).reset_index()
print(el)

plt.pie(el['Purchase_Amount'],labels=el['Education_Level'],startangle=140,autopct='%1.1f%%')
plt.show()
```
